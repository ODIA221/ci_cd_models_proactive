```python
# Cellule 1: Importations
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
from pathlib import Path

# Configuration des graphiques
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline

# Cellule 2: Chargement des données
data_path = Path("/../data/raw/metrics/")
# À adapter avec votre nom de fichier
#df = pd.read_csv(data_path / "compute_dataset.csv")  # MODIFIER ICI
#df = pd.read_csv('data/raw/dataset.csv')
# Chemin absolu (plus long mais toujours fiable)
df = pd.read_csv('/Users/Odia/Workspace/models_these/data/raw/metrics/dataset_metrics.csv')

print(f"Shape: {df.shape}")
print(f"\nColonnes: {df.columns.tolist()}")
print(f"\nTypes:\n{df.dtypes.value_counts()}")

# Cellule 3: Aperçu des données
df.head()

# Cellule 4: Statistiques descriptives
df.describe()

# Cellule 5: Valeurs manquantes
missing = df.isnull().sum()
missing_pct = 100 * missing / len(df)
missing_df = pd.DataFrame({'manquantes': missing, 'pourcentage': missing_pct})
missing_df[missing_df['manquantes'] > 0].sort_values('pourcentage', ascending=False)

# Cellule 6: Distribution des métriques clés
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

numerical_cols = df.select_dtypes(include=[np.number]).columns[:6]
for idx, col in enumerate(numerical_cols):
    axes[idx].hist(df[col].dropna(), bins=50, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'Distribution: {col}')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Fréquence')

plt.tight_layout()
plt.show()

# Cellule 7: Matrice de corrélation
plt.figure(figsize=(12, 10))
corr = df.select_dtypes(include=[np.number]).corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, cmap='coolwarm', center=0, 
            annot=False, square=True, linewidths=0.5)
plt.title('Matrice de corrélation des métriques CI/CD')
plt.tight_layout()
plt.show()

# Cellule 8: Détection des outliers (IQR)
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower) | (data[column] > upper)]
    return len(outliers), lower, upper

for col in numerical_cols[:5]:
    n_outliers, lower, upper = detect_outliers_iqr(df, col)
    print(f"{col}: {n_outliers} outliers ({100*n_outliers/len(df):.1f}%)")
```
